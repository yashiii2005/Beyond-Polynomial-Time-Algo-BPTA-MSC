generate the graph randomly, store it in a file, read the file, and then display the graph

```python
import random
import networkx as nx
import matplotlib.pyplot as plt
```

```python
vertices = random.randint(2, 12)
edges = vertices*(vertices-1)/2
file_name = "graph.txt"
```

```python
edgel=[]
i=0
while i <= int(edges):
  start = random.randint(1, vertices)
  end = random.randint(1, vertices)
  '''while start == end:
    end = random.randint(1, vertices)'''
  print(start,',',end)
  if [start,end] not in edgel :
    edgel.append([start,end])
    i+=1
```

```python
with open(file_name, "w") as file:
    for start,end in edgel:
        file.write(str(start) + " " + str(end) + "\n")
```

```python
graph = nx.DiGraph()
with open(file_name, "r") as file:
    for line in file:

        start, end = line.split()

        graph.add_edge(start, end)
```

```python
print ('vertices : ', vertices)
print ('edges : ', edges)
nx.draw(graph, with_labels=True)

plt.show()

#plt.savefig("graph.png")
```
